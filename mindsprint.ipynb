```python
data = {
    'customer_id':     [1, 2, 3, 4, 5, 6],
    'predicted_prob':  [0.90, 0.70, 0.40, 0.65, 0.30, 0.61],
    'actual_churned':  [1, 0, 1, 1, 1, 0]
}
#Compute precision and recall at a threshold of 0.6 (predicted positive if predicted_prob >= 0.6), without using sklearn's built-in functions.
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
threshold = 0.6

predected_positive = df['predicted_prob'] >= threshold
actual_positive = df['actual_churned'] == 1

tp = (predected_positive & actual_positive).sum()
fp = (predected_positive & ~actual_positive).sum()
fn = (~predected_positive & actual_positive).sum()

precision = tp +(tp+fp)
recall = tp + (tp +fn)

print(f"True Positives (TP): {tp}")
print(f"False Positives (FP): {fp}")
print(f"False Negatives (FN): {fn}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
```

```python
history = {
    "L1": ["Current", "Current", "30DPD", "60DPD", "90+DPD"],
    "L2": ["Current", "30DPD", "Current", "Current", "Closed"],
    "L3": ["30DPD", "30DPD", "60DPD", "30DPD", "Current"],
    "L4": ["Current", "Current", "Current", "Closed"],
}
states = ["Current", "30DPD", "60DPD", "90+DPD", "Closed"]

counts = {}
for s in states:
    counts[s] = {}
    for t in states:
        counts[s][t] = 0

for loans ,path in history.items():
    for i in range(len(path)-1):
        counts[path[i]][path[i+1]] += 1

prob = {}

for s in states:
    row_total = 0
    for t in states:
        row_total += counts[s][t]
    prob[s] = {}
    for t in states:
        prob[s][t] = round(counts[s][t] / row_total, 2) if row_total > 0 else 0.0

print(f"{'from/t0':<8}" + "".join(f"{t:>8}" for t in states))
for s in states:
    print(f"{s:<8}" + "".join(f"{prob[s][t]:<8.2f}" for t in states))

```
